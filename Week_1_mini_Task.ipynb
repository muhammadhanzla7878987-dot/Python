```python
import json
class Student:
    def __init__(self, name, marks):
        self.name = name
        self._marks = marks
    def grade(self):
        if self._marks >= 80:
            return "A"
        elif self._marks >= 60:
            return "B"
        elif self._marks >= 50:
            return "C"
        else:
            return "F"
class AIStudent(Student):
    def __init__(self, name, marks, ai_skill):
        super().__init__(name, marks)
        self.ai_skill = ai_skill
    def show(self):
        print(self.name, self._marks, self.grade(), self.ai_skill)
students = [
    AIStudent("Ali", 85, "Machine Learning"),
    AIStudent("Ahmed", 72, "Python"),
    AIStudent("Hassan", 91, "Deep Learning")
]
print("===== STUDENT MANAGEMENT SYSTEM =====")

for student in students:
    student.show()
search = input("\nEnter student name to search: ")
for student in students:
    if student.name.lower() == search.lower():
        print("Student Found:")
        student.show()
data = []
for student in students:
    data.append({
        "name": student.name,
        "marks": student._marks,
        "grade": student.grade(),
        "ai_skill": student.ai_skill
    })
with open("students.json", "w") as file:
    json.dump(data, file, indent=4)
print("\nData saved successfully in students.json")
```
